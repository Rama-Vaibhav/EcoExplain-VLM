# 04 — Freeze JSONL + 366-patch splits

| Action | Result |
|--------|--------|
| Rename previous `dataset.jsonl` | `dataset_old.jsonl` |
| Promote extensive analysis JSONL | `dataset.jsonl` (v2, unique Condition C text) |
| Stratified split | train 70% / dev 15% / test 15% on `candidate_class` |

CLI: `python scripts/05_freeze_jsonl_and_splits.py`


```python
from pathlib import Path
import importlib.util
import json
from collections import Counter

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent

spec = importlib.util.spec_from_file_location(
    "freeze05", ROOT / "scripts" / "05_freeze_jsonl_and_splits.py"
)
mod = importlib.util.module_from_spec(spec)
spec.loader.exec_module(mod)

mod.freeze_jsonl()
mod.write_splits(seed=42, train_frac=0.70, dev_frac=0.15)
```

```python
rows = [json.loads(l) for l in (ROOT / "dataset.jsonl").read_text().splitlines() if l.strip()]
print("dataset.jsonl rows", len(rows))
print("explanation_source", Counter(r.get("explanation_source") for r in rows))
print("unique prompts", len({(r.get("text") or {}).get("condition_c_user_prompt") for r in rows}))
print("old backup exists", (ROOT / "dataset_old.jsonl").exists())
print((ROOT / "data" / "splits" / "split_manifest.json").read_text())
```
